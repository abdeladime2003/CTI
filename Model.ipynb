```python
import pyspark
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
from pyspark.ml.feature import Tokenizer, HashingTF, StringIndexer, VectorAssembler
from pyspark.ml.classification import LogisticRegression
```

```python
from pyspark.ml import Pipeline
```

```python

from pyspark.ml.evaluation import MulticlassClassificationEvaluator
```

```python
# 1. Create Spark session
spark = SparkSession.builder \
    .appName("CSIC MLlib Training") \
    .config("spark.driver.memory", "4g") \
    .getOrCreate()
```

```python
# 2. Load CSV dataset into Spark DataFrame
df_ML = spark.read.csv("csic_train.csv", header=True, inferSchema=True)
```

```python
# 3. Select and clean useful columns
selected_cols = ["Method", "URL", "User-Agent", "content", "classification"]
df_ML = df_ML.select(*selected_cols).na.fill({"content": ""})

# ⚠️ NEW: Sample 5000 normal + 5000 anomalies
#normal_df = df_ML.filter(col("classification") == 0).orderBy(col("Method")).limit(5000)
#anomaly_df = df_ML.filter(col("classification") == 1).orderBy(col("Method"), ascending=False).limit(5000)
#df_sample = normal_df.union(anomaly_df)

# 4. Encode Method column
method_indexer = StringIndexer(inputCol="Method", outputCol="Method_Indexed")

# 5. Tokenize and vectorize text columns
url_tokenizer = Tokenizer(inputCol="URL", outputCol="URL_Tokens")
url_tf = HashingTF(inputCol="URL_Tokens", outputCol="URL_Features")

agent_tokenizer = Tokenizer(inputCol="User-Agent", outputCol="Agent_Tokens")
agent_tf = HashingTF(inputCol="Agent_Tokens", outputCol="Agent_Features")

content_tokenizer = Tokenizer(inputCol="content", outputCol="Content_Tokens")
content_tf = HashingTF(inputCol="Content_Tokens", outputCol="Content_Features")
```

```python
# 6. Assemble features
assembler = VectorAssembler(
    inputCols=["Method_Indexed", "URL_Features", "Agent_Features", "Content_Features"],
    outputCol="features"
)

```

```python
from pyspark.ml.classification import LinearSVC

# 7. Define model (SVM)
lr = LogisticRegression(labelCol="classification", featuresCol="features")
```

```python
# 8. Build pipeline
pipeline = Pipeline(stages=[
    method_indexer,
    url_tokenizer, url_tf,
    agent_tokenizer, agent_tf,
    content_tokenizer, content_tf,
    assembler,
    lr
])
```

```python
# 9. Fit the model
model = pipeline.fit(df_ML)

# 10. Predict on training data
predictions = model.transform(df_ML)

# 11. Evaluate model
evaluator = MulticlassClassificationEvaluator(
    labelCol="classification",
    predictionCol="prediction",
    metricName="f1"
)

f1_score = evaluator.evaluate(predictions)
print(f"✅ F1-score on training data: {f1_score:.4f}")
```

```python
# 12. Load test dataset (without labels)
df_test = spark.read.csv("test_dataset_10000.csv", header=True, inferSchema=True).na.fill({"content": ""})

# 13. Predict on test dataset
test_predictions = model.transform(df_test)
test_predictions.select("Method", "URL", "prediction").show(10)
```

```python
from pyspark.sql.functions import col, monotonically_increasing_id

# 14. Evaluate model on test set (with true labels)
df_labels = spark.read.csv("test_labels_10000.csv", header=True, inferSchema=True)

# Add an index to both DataFrames to join them
predictions_indexed = test_predictions.withColumn("index", monotonically_increasing_id())
labels_indexed = df_labels.withColumn("index", monotonically_increasing_id())

df_test_labeled = predictions_indexed.join(labels_indexed, on="index").drop("index")

# Evaluate
test_evaluator = MulticlassClassificationEvaluator(
    labelCol="classification",
    predictionCol="prediction",
    metricName="f1"
)

f1_score_test = test_evaluator.evaluate(df_test_labeled)
print(f"✅ F1-score on test data: {f1_score_test:.4f}")
```
